# **NumPy Indexing, Slicing and Strides:**

# **Indexing**

```python
arr=np.array([[20,30,40],
              [50,60,70],
              [80,90,100]])
print(arr[0,2])           # 40 (1st row,3rd column)
print(arr[1,1])           # 300 (2nd row, 2nd column)
```

```python
print(arr[-1,-2])         # 90 (last row , second last column)
print(arr[-1,-1])         # 100 (last row , last column)
```

```python
# Genral form: [ stat , stop , step ]
print(arr[0:2,0:1])
```

```python
dig_matrix=np.identity(5)
print(dig_matrix[::2,::2])  # (every 2nd row and 2nd column)
```

# **Slicing**

```python
slice_arr=arr[0,:2]
print(slice_arr)
```

```python
slice_arr[0]=10    # its view not copy so change original array.
print(arr[0][0])
```

```python
copy_arr=arr[0].copy  # copy of array
copy_arr=5
print(copy_arr)
```

```python
print(arr[0][0])      # unchanged original
```

# **Strides**
***Stride tells NumPy how far it needs to jump in memory to reach the next element along each dimension.***

```python
one_dim_arr=np.array([12,13,14,15,16])
print(one_dim_arr.strides)
```

```python
two_dim_arr=np.array([[3,6,9,12,15],
                      [4,8,12,16,20],
                      [5,10,15,20,25]])
print(two_dim_arr.strides)                # (40,8) move 40 bytes in each row because every row contaions 5 elements (8*5) and 8 bytes in each column.
```

```python
# change multi dim arr into 1D arr by returning it's copy
ar1=[[10,20,30,40],[15,25,35,45]]
ar1=np.array(ar1)
_ar1=ar1.flatten()      
_ar1[0]=5               # changing value at 0th index
print(_ar1)
print(ar1)              # change does not reflect at orginal arr due to flatten()
```

```python
# insert value
print(_arr_1D)
new_arr=np.insert(_arr_1D,1,9)    # insert(array,indx,value)
print(new_arr)
print(ar1)
new_ar1=np.insert(ar1,1,[100,200,300,400],axis=0)
print(new_ar1)
```

```python
# append value in arr:
updated_arr=np.append(_arr_1D,[9,10])    
print(_arr_1D)
print(updated_arr)
```

```python
# concatenation (union) of arr:
ar1=np.array([2,4,6,8])
ar2=np.array([10,12,14,16])
ar_sum=np.concatenate((ar1,ar2),axis=0)
print(ar_sum)
```

```python
# deletion of arr:
new_ar_sum=np.delete(ar_sum,-1)
print("before deletion: ")
print(ar_sum)
print("after deletion:")
print(new_ar_sum)
_arr_2D=np.delete(arr_2D,0,axis=0)
print("before deletion:")
print(arr_2D)
print("after deletion:")
print(_arr_2D)
```

```python
# vertically stack:
print(ar1)
print(ar2)
new_ar=np.vstack((ar1,ar2))
print(new_ar)
```

```python
# horizontally stack:
print(ar1)
print(ar2)
new_ar=np.hstack((ar1,ar2))
print(new_ar)
```

```python
# split an arr:
new_ar1=np.split(ar1,2)
print(new_ar1)
```

```python
# broadcasting:
prices=np.array([100,200,300,400])
discount=10/100
final_prices=prices-prices*discount
print(final_prices)
```

```python
# vectorization:
ar3=np.array([10,20,30])
ar4=np.array([100,200,300])
res=ar3+ar4
multiplied=ar4*2
print(res)
print(multiplied)
```

```python
# missing values:
_arr2=np.array([1,np.nan,2,3,4,np.nan])
print(np.isnan(_arr2))    # return true at the places where missing value occur.
```

```python
# replace missing values with 100.
cleaned_arr2=np.nan_to_num(_arr2,nan=100)   
print(cleaned_arr2)
```
