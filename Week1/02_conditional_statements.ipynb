# Conditional statements:
conditional statements are used to define conditions for code.

There are three types of conditions used:

✔ **if**

✔ **elif**

✔ **else**

```python
temp=float(input("enter temperature:"))
if temp>35:
    print("its hot!")
else:
    print("its not hot")
```

```python
age=int(input("enter your age:"))
if age>=18:
    print("can vote and drive")
elif age>10:
    print("can drive")
else:
    print("you are a child")
        
```

```python
light=input("enter traffic light")
if light=="red":
    print("wait")
elif light=="green":
    print("Go")
elif light=="yellow":
    print("look")
else:
    print("light broken")
    
```

```python
marks=int(input("enter marks:"))
if marks>=90:
    print("A+ Grade")
elif marks>=80:
    print("A Grade")
else:
    print("Fail")
```

```python
age=int(input("enter your age:"))
country=input("enter your country:").capitalize()
if age>=18 and country =="Pakistan":
    print("You can vote in Pakistan")
else:
    print("you are not eligable for vote.")
```
