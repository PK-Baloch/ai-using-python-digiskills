# What is a List?

A list is an ordered, changeable collection.

✔ **Ordered**  
✔ **Mutable (can change)**  
✔ **Allows duplicates**

```python
fruites=["Apple","Banana","Banana","Mango"]
```

```python
print(fruites)
```

```python
print(fruites[2])
```

```python
print(fruites[-1]) 
```

```python
print(fruites)
```

```python
fruites.insert(1,"Grapes")  # add item at specific index
```

```python
print(fruites[1])
```

```python
fruites.remove("Banana")    # remove item 
```

# What is a Dictionary ?
It is a python collection in which values are stored in "key:value" pairs.

Dictionaries are:

✔ **Unordered**

✔ **Mutabale**

✔ **Don't allow duplicate keys**

```python
Student={
    "Name":"Aqib Ali",
    "Subjects":{
        "Math":84,
        "chem":74,
        "Phys":83}}
```

```python
print(Student)
```

# Access values:

```python
print(Student["Subjects"]["Math"]) # we can access particular vaue in a dictionary
```

```python
print(Student["Subjects"])
```

```python
print(Student.get("Name"))
```

# Operators:
Operators are symbols or keywords used to perform operations on values and variables.

✔ **Arithamatic Operators**

✔ **Logical Operators**

✔ **Assignment Operators**

✔ **Relational Operators**

```python
# 1. Arithamatic Operators: used to perform arithamatic operations like +, - , /, * 
num1=50
num2=25
print(num1+num2)    # adds both numbers
print(num1*num2)    # multiply both bumbers
print(num1-num2)    # substract num2 from num1
print(num1//num2)   # devide num1 with num2
print(num1%num2)    # give reminder
print(num1**2)      # power of num1 
```

```python
# 2. Logical Operators: we have three logical operators AND, OR , NOT
print(not(num1>num2))               # False
print(not(num1<num2))               # True
print(not(num1>num2))               # False
print((num1>num2)or(num1>=num2))    # True
print((num1>0) and (num2<30))       # True
```

```python
# 3. Relational Operators: used to compare two values 
print(num1==num2)   # False
print(num1<num2)    # False
print(num1>num2)    # True
print(num1>=num2)   # True
print(num1<=num2)   # False
```

```python
# 4. Assignment Operators: used to assign values to a variable
num1=44
print(num1)
num1+=num1
print(num1)
num1*=num1
print(num1)
```
