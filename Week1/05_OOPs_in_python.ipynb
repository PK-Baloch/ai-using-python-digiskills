# **Getters and Setters**

# **Getters:**
***Geters are those methods which are used to retrieve the data attributes of the class.***
# **Setters**
***Setters are the functions used to assign or modify the data attributes of the class.***

```python
class Employ:
    def __init__(self,name,sal):
        self.name=name
        self._sal=sal
    @property
    def get_sal(self):
        return self._sal
    def set_sal(self,sal):
        if sal>=0:
            self._sal=sal
        else:
            print("Invalid sal input")
e1=Employ("Raza M",34000)
e1.set_sal(23000)
print(e1.get_sal)
```

# **2.Abstraction:**

# **Example:**

```python
class Car:
    def __init__(self):
        self.accelerator=False  # hiding internal machanism of car 
        self.clutch=False
        self.Break=False
    def start(self):
        self.accelerator=True
        self.clutch=True
        print("Car started")
car1=Car() 
car1.start()                   # Giving only access to usefull features
```

# **3.Inheritance:**
***when a child/drived class drives or inherit the properties from it parent/base class then this is called inheritance.***
# **Example:**

```python
class Animal:
    def speak(self):
        print("any sound")
class Lion(Animal):
    pass
l1=Lion()
l1.speak()
```

# **1. Single Inheritance:** 
***Inheritance in which a parent class inherit thier properties to child class.***

# **Example:** 




```python
class Tree:
    def show_basic_prop(self):
        print("Take carbon di oxide")
        print("Release Oxygen")
        print("Make food through Photo synthesis")
class Neem_tree(Tree):
    def properties(self):
        print("Neem leaves are good for skin")
        print("Neem oil is used for hair care")
n1=Neem_tree()
n1.show_basic_prop()
n1.properties()
```

# **2. Multi Level Inheritance:**
***In this type of inheritance parent class inherit their properties to child class and this child class also inherit their properties to other sub child class***

# **Example:**

```python
class Car:
        def start(self):
            print("Car is Started")

class ToYoTa(Car):
    def __init__(self,brand):
        self.brand=brand
class Fortuner(ToYoTa):
    def __init__(self, type):
        self.type=type

Car1=Fortuner("Diesel")
Car1.start()
```

# **3. Multiple Inheritance:**
***Inheritance in which one parent class inherit thier properties to multiple child classes.***
# **Example:**

```python
class Person:
    def __init__(self,name,age,hight):
        self.name=name
        self.age=age
        self.hight=hight
class Teacher(Person):
    def __init__(self,name,age,hight,sal,dept):
        super().__init__(name,age,hight)       # means run my parent class's constructor.
        self.sal=sal
        self.dept=dept
    def show_info(self):
        print(f"name:{self.name}")
        print(f"age:{self.age}")
        print(f"sal:{self.sal}")
        print(f"dept:{self.dept}")
t1=Teacher("Saifullah",43,5.3,240000,"IT")
t1.show_info()
```

# **Method Overriding**

```python
class Animal:
    def speak(self):
        print("some sound")
class Dog(Animal):
    pass
    def speak(self):                # override method.
        print("Wolf! wolf! wolf!")
d1=Dog()
d1.speak()
```

# **Special Methods:**
***(Dunder Functions)***

```python
# __str__()
class Book:
    def __init__(self,title,auther):
        self.title=title
        self.auther=auther
    def __str__(self):
        return f"'{self.title}' by {self.auther}"
b1=Book(1998,"George Orwell")
print(b1)
        
```

```python
# __len__()
class Playlist:
    def __init__(self,songs):
        self.songs=songs
    def __len__(self):
        return len(self.songs)
    def __add__(self,others):
        return Playlist(self.songs+others.songs)
p1=Playlist(["song1","song2","song3"])
p2=Playlist(["song4"])
print(len(p1))
print(len(p1+p2))
```

```python
class Circle:
    def __init__(self,raduis):
        self.raduis=raduis
    def area(self):
        return (22/7)*self.raduis*self.raduis
    def perimeter(self):
        return 2*(22/7)*self.raduis
    
c1=Circle(6)
print(c1.area())
print(c1.perimeter())
```

***4.Create a Complex class and define dunder functions to add and sub complex numbers.***

```python
class Complex:
    def __init__(self,real,img):
        self.real=real
        self.img=img
    def showNumber(self):
        print(self.real,"i +" ,self.img,"j")
    def __add__(self,num2):         # dunder function
        New_real=self.real+num2.real
        New_img=self.img+num2.img
        result=Complex(New_real,New_img)
        return result
    def __sub__(self,num2):         # dunder function
        New_real=self.real-num2.real
        New_img=self.img-num2.img
        result=Complex(New_real,New_img)
        return result
num1=Complex(2,4)
num1.showNumber()
num2=Complex(1,3)
num2.showNumber()
num3=num1+num2
num3.showNumber()
num3=num1-num2
num3.showNumber()
```
