# **What is a loop:**
A loop is used to repeat instrructions in coding.

There are mainlly two types of loops used in python.

✔ **For**

✔ **While**


```python
for i in range(3):
    print("Python is fun")
```

```python
for char in "python":
    print(char)
```

```python
i=4
row=i
col=i
matrix=[]
for r in range(row):
    row=[]
    for c in range(col):
        row.append(0)
    matrix.append(row)
for row in matrix:
    print(' '.join(map(str,row)))
    
```

```python
i=1
while i<=5:
    print(i)
    i+=1
```

# Nested loop:

```python

for k in range(3):
    print("Outer loop")
    for j in range(3):
        print("    Inner loop")
```

# BREAK & CONTINUE:

# ***BREAK***
used to terminate the loop when encountered.

```python
colors=["red","white","blue","green","yellow","grey"]
for i in range(len(colors)):
    if colors[i]=="blue":
        print("color found at index: ",i)
        break
    else:
        print("finding...")
```

# ***Continue*** 
terminates the execution in the current iteration and continues execution of the loop with next iteration.

```python
# skip even numbers and print odd numbers.
for i in range(20):
    if i%2==0:
        continue
    else:
        print(i)
        
```

```python
# find the sum of first n numbers
n=int(input("enter the number:"))
sum=0
for i in range(1,n+1):
    sum+=i
print(sum)
    
    
```

```python
# find fact of n nums.
num=int(input("enter a number:"))
fact=1
while num>=1:
    fact*=num
    num-=1
print(fact)
```
